```python
import os

import matplotlib.pyplot as plt
import scipy.sparse as sp
import scanpy as sc
import harmonypy as hm
import lamindb as ln

ln.track("nZvPcHEEtXr0")
```

```python
SWEEP_DIR   = '/home/workspace/PCD_data/xenium/commercial/data/resolution_sweep'
FIGURES_DIR = '/home/workspace/PCD_data/xenium/commercial/figures/processing'

for directory in (SWEEP_DIR, FIGURES_DIR):
    os.makedirs(directory, exist_ok=True)
```

```python
def save_fig(fig, stem, description, ulabels=(), features=None):
    fig.tight_layout()
    for ax in fig.axes:
        for coll in ax.collections:
            coll.set_rasterized(True)
    for ext in ['png', 'pdf']:
        path = os.path.join(FIGURES_DIR, f'{stem}.{ext}')
        fig.savefig(path, dpi=150, bbox_inches='tight')
        artifact = ln.Artifact(path, key=f'test_{stem}.{ext}', description=description).save()
        if ulabels:
            artifact.ulabels.add(*ulabels)
        if features:
            artifact.features.add_values(features)
    plt.close(fig)
```

```python
combined = ln.Artifact.get('Po61viBGaIukVXxG0001').load()

fig = sc.pl.umap(combined, color='sample', legend_loc='right margin',
                 show=False, return_fig=True, size=6)

plt.show()

save_fig(fig, key='fig2d_sample_umap',
         description='Figure 2D: UMAP colored by sample with legend on right margin')
```
